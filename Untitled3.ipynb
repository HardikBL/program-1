```python
import matplotlib.pyplot as plt
import csv

x = []
y = []

with open("stock_market.csv", "r") as csvfile:
    lines = csv.reader(csvfile, delimiter=",")

    for row in lines:
        x.append(row[0])
        y.append(float(row[1]))

print(x)
print(y)

plt.plot(x, y, color="darkblue", linestyle="-",
         marker="*", ms=15, mec="pink",
         label="Closing Price")

plt.xticks(rotation=45)
plt.xlabel("Date", fontsize=14, color="brown")
plt.ylabel("Closing Price", fontsize=14, color="brown")
plt.title("Stock Market Closing Price Trend",
          fontsize=20, color="g")

plt.grid()
plt.legend()
plt.show()
```
